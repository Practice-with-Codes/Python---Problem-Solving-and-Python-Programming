# Program 2

```python
# function to find the absolute value of the number

def absolutevalue(number):
    if number >= 0:
        return number
    else:
        return -number
    
print(absolutevalue(8))
print(absolutevalue(-14))
```

# Program 3

```python
# function to find the given number is odd or even

def odd_even(number):
    if number % 2 == 0:
        print("The number",  number , "is even")
    else:
        print("The number", number , "is odd")
    return;

result = int(input("Enter a number: "))
odd_even(result)
```

# Pass by Reference Vs Pass by Value

# Program 1 - Pass by Reference

```python
def change(list):
    """It changes the list which was already passed to this function"""
    list.append([5, 6, 7, 8])
    print("The numbers inside the function: ", list)
    
# call the function change
list = [150, 250, 350]
change(list);
print("The numbers outside the function: ", list)
```


# Program 2 - Pass by Value

```python
def change(list):
    """It changes the list which was already passed to this function"""
    list = [5, 6, 7, 8]
    print("The numbers inside the function: ", list)
    
# call the function change
list = [150, 250, 350]
change(list);
print("The numbers outside the function: ", list)
```

# Program - Python Default Arguments

```python
def greeting(person_name, message = "Good Day!"):
    """The function greeting the person with the given message. If there is no message it takes 
    the message Good Day! by default"""
    print("Hai", person_name + ',' + message)
    
greeting("Sham")
greeting("Varun", "How are you?")
```

# Program - Python Keyword Arguments

```python
# a function with two keyword arguments

greeting(person_name = "Varun", message = "How are you?")
```

```python
# function without order arguments

greeting(message = "How are you?", person_name = "Varun")
```

```python
# function with one positional arguments and one keyword argument

greeting("Varun", message = "How are you?")
```

```python
# Error

greeting(person_name = "Varun" , "How are you?")
```

# Program - Required Argumnents

# Program - Python Variable Length(Arbitrary) Argumnents

```python
def greeting(*person_names):
    """It greets every person in the person_names tuple"""
    # person_names is a tuple with arguments
    print("Hai", person_names)
    
greeting("Anu", "Kavin", "Ram", "Krish")    
```
