# Composition

```python
"""Function Composition: 
   Combining function such that the result of each function is passed as an argument of the next function """
```

```python
def compose1(f, g):
    return lambda x: f(g(x))
def double(x):
    return x * 2
def inc(x):
    return x + 1

inc_and_double = compose1(double, inc)
inc_and_double(10)
```

```python
# composing n functions
# compare more than two functions
# Example: Compose three functions --> compose first two functions, then compose the newly created function with the next one

def dec(x):
    return x - 1

inc_double_and_dec = compose1(compose1(dec, double), inc)
inc_double_and_dec(10)
```

```python
import functools            # functools --> higher-order function that act on or return other functions

def compose(*functions):
    return functools.reduce(lambda f, g: lambda x: f(g(x)), functions, lambda x: x)
    
inc_double_and_dec = compose(dec, double, inc)
inc_double_and_dec(10)
```
