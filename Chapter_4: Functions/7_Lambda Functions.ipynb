# Lambda Functions

```python
# A function is anonymous when it is defined without a name
# Normal functions defined using def keyword
# In Python, anonymous function defined using lambda keyword
```

```python
# lambda functions can have any number of arguments, but only one expression
# Lambda functions can be used, wherever function objects are required
```

```python
# Program
# number --> argument
# number * 2 --> expression

value = lambda number: number * 2
print(value(25))
```

```python
def value(number):
    return number * 2

print(value(2))
```

```python
# Use: 
# when we require a nameless function for a short period of time
# used along with built-in functions like filter(), map()
```

```python
# lambda function with filter()
# Program: to filter even numbers from the given list

old_list = [21, 25, 24, 27, 28, 31, 33, 42]
current_list = list(filter(lambda number:(number % 2 == 0), old_list))
print(current_list)
```

```python
# lambda function with map()
# Program: to double all the items in the list

old_list = [21, 25, 24, 27, 28, 31, 33, 42]
current_list = list(map(lambda number:(number ), old_list))
print(current_list)
```
