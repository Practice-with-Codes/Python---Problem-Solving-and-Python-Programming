# AD__14: Using Recursion Function Find the Sum of n Natural Numbers

```python
def recursive_sum(number):
    """Recursive function to return the sum of n natural numbers"""
    if number <= 1:
        return number
    else:
        return number + recursive_sum(number - 1)
# changing the value to produce the diverse result
value = 42
# Getting the input
# value = int(input("Enter the value: "))
if value < 0:
    print("Enter the positive number")
else:
    print("The result is ", recursive_sum(value))
```
