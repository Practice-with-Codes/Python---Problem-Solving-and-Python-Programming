# AD_7: Program to Design an Arithmetic Calculator

```python
def multiplication(value1, value2):
    """Function to calculate the multiplication"""
    return value1 * value2
def division(value1, value2):
    """Function to calculate the division"""
    return value1 / value2
def subtraction(value1, value2):
    """Function to calculate the subtraction"""
    return value1 - value2
def addition(value1, value2):
    """Function to calculate the addition"""
    return value1 + value2
print("Choose the required operation:")
print("1. Addition")
print("2. Subtraction")
print("3. Multiplication")
print("4. Division")
option = input("Choose the option (1/2/3/4): ")
number1 = int(input("Enter the first number: "))
number2 = int(input("Enter the second number: "))
if option == '1':
    print(number1, "+" , number2 , "=" , addition(number1, number2))
elif option == '2':
    print(number1, "-" , number2 , "=" , subtraction(number1, number2))
elif option == '3':
    print(number1, "*" , number2 , "=" , multiplication(number1, number2))
elif option == '4':
    print(number1, "/" , number2 , "=" , division(number1, number2))
else:
    print("Invalid Input")
```
