# AD_6: Program to Find the Least Common Multiple of a Number

```python
def calculateLCM(number1, number2):
    if number1 > number2:
        great = number2
    else:
        great = number1
    while(True):
        if((great % number1 == 0) and (great % number2 == 0)):
            result = great
            break
            great += 1
    return result
val1 = 36
val2 = 40
print("The least common multiple of ", val1 , "and", val2, "is", calculateLCM(val1, val2))
```
