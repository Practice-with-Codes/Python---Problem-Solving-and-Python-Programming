# AD_9: Program using Functions to Find the Prime Numbers

```python
prime_nos = 0

def is_prime_no(x):
    if x < 2:
        return False
    for y in range(2, x):
        if x % y == 0:
            return False
    return True

count = int(input("How many numbers you wish to check? "))

for i in range(count + 1):
    if is_prime_no(i):
        prime_nos += 1
        print(i)

print("We found " + str(prime_nos) + " prime numbers")
```
