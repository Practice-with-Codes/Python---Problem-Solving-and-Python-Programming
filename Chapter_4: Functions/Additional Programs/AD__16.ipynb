# AD__16: Generate new string from an existing string by adding "Is" in the front, without modifying it

```python
def check(str1):
    if len(str1) >= 2 and str1[:2] == "Is":
        return str1
    return "Is " + str1
# Main program
str2 = input("Enter the string: ")
print("The updated string is: ", check(str2))
```
