# AD__15: Recursive Function to Find Factors of a Number

```python
def factors(number):
    # find the factors of a given number
    print("The factors of a given number", number , "are: ")
    for count in range(1, number + 1):
        if number % count == 0:
            print(count)
# changing the number to get a diverse result
value = 450
# value = int(input("Enter the number: "))
(factors(value))
```
