# AD_4: Program to Search an Element in an Array using Linear Search Technique

```python
# Program to search an element in an array using linear search technique

list = [4, 1, 2, 5, 3]                                       # set up an array
search = int(input("Enter the search number: "))             # enter the number to search
for i in range(0, len(list)):                                # repeat for each item in list
    if search == list[i]:                                    # if item at position i is search time
        print(str(search) + " found at position " + str(i))  # found
```
