# Python Recursion

```python
# Recursion --> function calls itself one or more times in its body
# returning the return value of the function call
```

```python
# recursive function to find the factorial of an integer
# recursively call itself by decreasing the number
# recursion ends when the number reduces to 1

def factorial(number):
    """To find the factorial of a given integer"""
    if number == 1:
        return 1
    else:
        return(number * factorial(number -1))
value = 3
print("The factorial of an integer ", value , "is", factorial(value))
```

```python
# Advantages 
# 1. code looks neat and elegant
# 2. compound assignment can be broken down into simpler sub-problems
# 3. Sequence generation is easier than nested iteration
```

```python
# Disadvantages
# 1. logic of recursion is rigid
# 2. recursive calls are exclusive as they require lot of memory and time
# 3. recursive functions are rigid to debug
```
